```python
# conditional statement

age=18
if age>18:
    print("you are eligible for voting")

elif age==18:
    print("you are eligible for voting but you have to wait for your birthday")

elif age<18:
    print("you are not eligible for voting")

else:
    print("invalid age")
```

```python
#looping statement 
#for loop and while loop

list1=[1,2,3,4,5]
for x in list1:
    print(x)

for x in range(4,10,2):    #parameter start 0, stop n-1, step 1
    print(x)

    #while loop
i=1

while i<=10:
    print(i)
    i+=1
```

```python
# create a program where the user will enter a number and the program will
# add all the numbers one by one and print the result until
# the user will enter Q to Quit The program
# hint while loop and break statement , start with sum=0 and add the number to sum 
```

```python
sum = 0

while True:
    user = input("Enter a number (or Q to Quit): ")

    if user.upper() == "Q":
        break

    sum += int(user)
    
print("Total =", sum)
```

```python
# Jumping statement
# continue statement
for i in range(1,11):
    if i==5:
        continue
    print(i)
```

```python
# break statement
for i in range(1,11):
    if i==5:
        break
    print(i)
```
