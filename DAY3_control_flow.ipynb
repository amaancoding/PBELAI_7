```python
# conditional statement

while True:
    age_input = input("enter your age: ").strip()
    if age_input == "":
        print("Please enter your age.")
        continue
    try:
        age = int(age_input)
        break
    except ValueError:
        print("Invalid input. Enter a valid integer.")
if age>18:
    print("you are eligible for voting")

elif age==18:
    print("you are eligible for voting but you have to wait for your birthday")

elif age<18:
    print("you are not eligible for voting")

else:
    print("invalid age")
```
