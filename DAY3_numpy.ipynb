```python
n=20 # the number to guess

add_num = 0   # times guess was too small (need to add)
subs_num = 0  # times guess was too big (need to subtract)

print("Guess the number:")

while True:
    guess = int(input("Enter your guess: "))

    if guess > n:
        print("big! Try a smaller number.")
        subs_num += 1
    elif guess < n:
        print("Too small! Try a bigger number.")
        add_num += 1
    else:
        print("Correct! You guessed the number.")
        break

print("\nTotal times you guessed too small:", add_num)
print("Total times you guessed too big:", subs_num)
```
