```python
# create a guessing game where the user willenter a number and you have to tell number is big or small
# loop user will enter another number make it untill user guess the number 
# add num , subs num

import random

num = random.randint(1, 100)
count = 0

while True:
    guess = int(input("Guess a number (1-100): "))
    count += 1

    if guess > num:
        print("Number is big!")
    elif guess < num:
        print("Number is small!")
    else:
        print("Correct! You guessed it in", count, "attempts.")
        break
```
